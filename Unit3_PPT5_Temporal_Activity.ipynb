## Part B — The vendor's sampling proposal (5 min) 
The vendor wants to record **one sample every 25 hours**. Traffic has a strong **24-hour cycle**.

**B1.** Implement:
- `nyquist_max_interval(period_h)`: the **largest** sampling interval that avoids aliasing (Nyquist: $f_s > 2f$)
- `aliased_period(true_period, sample_interval)`: the **apparent period**, using aliased frequency $f_a = |f - f_s|$ for $f_s/2 < f < 3f_s/2$ (the case in this activity)

Then compute `apparent_period_25h` for the vendor's plan.

```python
def nyquist_max_interval(period_h):
    return period_h / 2

def aliased_period(true_period, sample_interval):
    f, fs = 1 / true_period, 1 / sample_interval
    fa = abs(f - fs)
    return 1 / fa

apparent_period_25h = aliased_period(24, 25)

check_B(nyquist_max_interval, aliased_period, apparent_period_25h)
```

```python
# (given) What the vendor's data would look like for spring 2017
seg = tr["traffic_volume"].loc["2017-03-01":"2017-05-31"].asfreq("h").interpolate(limit=3)
fig, ax = plt.subplots(figsize=(14, 3.5))
ax.plot(seg.index, seg.values, c="lightgrey", lw=0.4, label="true hourly")
ax.plot(seg.index[::25], seg.values[::25], "-o", c="red", ms=3, label="vendor: every 25 h")
ax.set_title(f"Sampling every 25 h turns the 24-h cycle into a fake ~{apparent_period_25h/24:.0f}-day wave"); ax.legend(); plt.show()
```

**B2. A manager looks only at the vendor's 25-hour samples and says: *'Traffic rises and falls on a roughly 25-day cycle — let's schedule closures in the low phase.'* Explain what is really happening, using the words sampling rate, Nyquist and aliasing.**

*Your answer:*
The manager is observing severe **aliasing** caused by an inadequate **sampling rate** that violates the **Nyquist** criterion. Because the true traffic variation has a fundamental 24-hour cycle, Nyquist requires sampling at least twice per cycle ($f_s > 2f$, so $\Delta t < 12$ hours). Sampling every 25 hours shifts each consecutive sample 1 hour later in the daily cycle, causing the high-frequency daily peak to beat against the sampling rate and fold into a phantom low-frequency wave with an apparent period of $|1/24 - 1/25|^{-1} = 600\text{ hours} = 25\text{ days}$. The apparent 25-day cycle is an entirely synthetic artifact of stroboscopic undersampling, not a physical traffic phenomenon.

**B3. What maximum sampling interval would you recommend to MnDOT? Would sampling exactly every 12 hours be safe? Justify briefly.**

*Your answer:*
I would recommend a maximum sampling interval of at most 1 hour (or 15–30 minutes) to capture morning/evening rush hour peaks and sharp transitions. Sampling at exactly 12 hours is the theoretical Nyquist limit for a pure 24-hour sine wave, but it is strictly unsafe: real traffic has higher-harmonic components (sharp 2-hour rush hour surges requiring $\Delta t < 1$ hour), and sampling at exactly $\Delta t = 12$ hours can hit zero-crossings or constant nodes, completely masking rush-hour peaks or introducing pathologically aliased amplitudes.

## Part E — Design the animated *Traffic Pulse* for the lobby screen (6 min) 
**Design brief:** the screen animates **24 hourly states** of a typical weekday. Each hour is shown for **0.75 s** at **30 fps**, with a smooth transition between hours.

💻 **E1.** Implement and compute:
- `frame_budget_ms(fps)`: milliseconds available per frame
- `smoothstep(u)`: the ease-in-out function $3u^2-2u^3$
- `n_frames`, `duration_s`: for the brief above
- `compression`: seconds of **data** per second of **animation** (one hour of data per 0.75 s)

```python
def frame_budget_ms(fps):
    return 1000.0 / fps

def smoothstep(u):
    return 3.0 * (u**2) - 2.0 * (u**3)

duration_s  = 24 * 0.75
n_frames    = int(duration_s * 30)
compression = 3600.0 / 0.75

print(f"Frame budget at 30 fps: {frame_budget_ms(30):.2f} ms")
print(f"Smoothstep at u=0.25: {smoothstep(0.25):.5f}")
print(f"Duration: {duration_s:.1f} s | Frames: {n_frames} | Compression: {compression:.0f}:1")
```

**E3. Play the animation and try to answer: *In which month is the 17:00 peak lowest?* Then answer it with the small multiples. Which was faster and more certain, and why? Use the terms working memory and 'eyes beat memory'.**

*Your answer:*

**E4. What would go wrong if you left out `range_y` (letting each frame auto-scale)? Which Heer & Robertson principle does that violate?**

*Your answer:*
Leaving out `range_y` causes each frame to auto-scale its vertical axis to its local monthly maximum, making a low-volume winter month look visually identical in height to a high-volume summer month and causing the y-axis ticks to jitter chaotically. This violates Heer & Robertson's principle of **Congruence** (visual changes must correspond directly to data changes, so a consistent spatial scale is essential for physical fidelity) and disrupts perceptual consistency.

**E5. Why use smoothstep (slow-in/slow-out) between hours instead of linear motion or a jump cut? Refer to your `smoothstep(0.25)` value and the per-frame jumps at the start and end.**

*Your answer:*
Smoothstep provides ease-in and ease-out transitions where velocity starts and ends at zero ($\frac{ds}{du}\big|_{u=0,1} = 0$), preventing jarring velocity discontinuities. At $u=0.25$, `smoothstep(0.25) = 0.15625` (substantially less than the 0.25 linear progression), demonstrating how the motion accelerates gradually from rest and decelerates smoothly toward the target. This gradual deceleration allows the human visual tracking system to fixate cleanly on incoming target values without overshoot, whereas linear transitions cause abrupt velocity shocks and jump cuts trigger visual masking / change blindness.
